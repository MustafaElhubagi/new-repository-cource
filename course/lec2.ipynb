```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\New folder\image_with_shapes_2.jpeg"
)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Detect edges
edges = cv2.Canny(gray, 50, 150)

# Find contours
contours, _ = cv2.findContours(
    edges,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# Copy image to draw results
result = image.copy()

for contour in contours:

    # Approximate the contour
    epsilon = 0.04 * cv2.arcLength(contour, True)
    approx = cv2.approxPolyDP(contour, epsilon, True)

    # Area of shape
    area = cv2.contourArea(contour)

    # Ignore very small objects
    if area < 500:
        continue

    # Center of shape
    M = cv2.moments(contour)

    if M["m00"] != 0:
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
    else:
        continue

    # Detect triangle
    if len(approx) == 3:
        shape = "Triangle"

    # Detect square
    elif len(approx) == 4:
        x, y, w, h = cv2.boundingRect(approx)
        ratio = w / float(h)

        if 0.90 <= ratio <= 1.10:
            shape = "Square"
        else:
            shape = "Rectangle"

    # Detect circle
    else:
        perimeter = cv2.arcLength(contour, True)

        if perimeter != 0:
            circularity = 4 * 3.14159 * area / (perimeter * perimeter)

            if circularity > 0.75:
                shape = "Circle"
            else:
                shape = "Unknown"
        else:
            shape = "Unknown"

    # Draw contour
    cv2.drawContours(result, [approx], -1, (0, 255, 0), 2)

    # Put shape name
    cv2.putText(
        result,
        shape,
        (cx - 40, cy),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 0, 255),
        2
    )

# Convert BGR to RGB
result = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)



plt.imshow(result)
plt.title("Shape Detection")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\New folder\image_with_shapes_2.jpeg"
)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Edge detection
edges = cv2.Canny(gray, 50, 150)

# Find contours
contours, _ = cv2.findContours(
    edges,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# Copy image
result = image.copy()

# Counters
triangle_count = 0
square_count = 0
rectangle_count = 0
circle_count = 0

# Detect shapes
for contour in contours:

    # Calculate area
    area = cv2.contourArea(contour)

    # Ignore small objects
    if area < 500:
        continue

    # Calculate perimeter
    perimeter = cv2.arcLength(contour, True)

    # Approximate contour
    epsilon = 0.04 * perimeter
    approx = cv2.approxPolyDP(contour, epsilon, True)

    # Calculate center
    M = cv2.moments(contour)

    if M["m00"] == 0:
        continue

    cx = int(M["m10"] / M["m00"])
    cy = int(M["m01"] / M["m00"])

    # Detect shape
    if len(approx) == 3:
        shape = "Triangle"
        triangle_count += 1

    elif len(approx) == 4:
        x, y, w, h = cv2.boundingRect(approx)
        ratio = w / float(h)

        if 0.90 <= ratio <= 1.10:
            shape = "Square"
            square_count += 1
        else:
            shape = "Rectangle"
            rectangle_count += 1

    else:
        circularity = 4 * 3.14159 * area / (perimeter * perimeter)

        if circularity > 0.75:
            shape = "Circle"
            circle_count += 1
        else:
            continue

    # Draw contour
    cv2.drawContours(
        result,
        [approx],
        -1,
        (0, 255, 0),
        2
    )

    # Display shape name
    cv2.putText(
        result,
        shape,
        (cx - 50, cy - 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (0, 0, 255),
        2
    )

    # Display area
    cv2.putText(
        result,
        f"Area: {area:.0f}",
        (cx - 50, cy),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.5,
        (255, 0, 0),
        2
    )

    # Display perimeter
    cv2.putText(
        result,
        f"Perimeter: {perimeter:.0f}",
        (cx - 70, cy + 25),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.5,
        (255, 100, 0),
        2
    )

# Total number of shapes
total_shapes = (
    triangle_count
    + square_count
    + rectangle_count
    + circle_count
)

# Print results
print("Number of triangles:", triangle_count)
print("Number of squares:", square_count)
print("Number of rectangles:", rectangle_count)
print("Number of circles:", circle_count)
print("Total shapes:", total_shapes)

# Convert BGR to RGB
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# Display result
plt.figure(figsize=(10, 7))
plt.imshow(result_rgb)
plt.title("Shape Detection")
plt.axis("off")
plt.show()
```

```python

import cv2
import matplotlib.pyplot as plt

# -------------------------------------------------
# Read image
# -------------------------------------------------
image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\New folder\bloodcells.jpeg")

if image is None:
    print("Error: Image not found!")
    exit()

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


# -------------------------------------------------
# 1. Threshold for DARK cells
# -------------------------------------------------
# الخلايا الداكنة تظهر كأبيض بعد Binary Inverse
_, dark_mask = cv2.threshold(
    gray, 100, 255, cv2.THRESH_BINARY_INV
)

# إزالة الضوضاء الصغيرة
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
dark_mask = cv2.morphologyEx(
    dark_mask,
    cv2.MORPH_OPEN,
    kernel
)


# -------------------------------------------------
# 2. Threshold for LIGHT / PALE cells
# -------------------------------------------------
# يمكنك تغيير 180 حسب الصورة
_, pale_mask = cv2.threshold(
    gray, 180, 255, cv2.THRESH_BINARY_INV
)

pale_mask = cv2.morphologyEx(
    pale_mask,
    cv2.MORPH_OPEN,
    kernel
)


# -------------------------------------------------
# Function to count cells
# -------------------------------------------------
def count_cells(mask, min_area=100):

    num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
        mask,
        connectivity=8
    )

    count = 0
    centers = []

    for i in range(1, num_labels):  # 0 = background

        area = stats[i, cv2.CC_STAT_AREA]

        # تجاهل الأجسام الصغيرة
        if area >= min_area:

            count += 1

            cx, cy = centroids[i]
            centers.append((int(cx), int(cy)))

    return count, centers


# -------------------------------------------------
# Count dark cells
# -------------------------------------------------
dark_count, dark_centers = count_cells(
    dark_mask,
    min_area=100
)


# -------------------------------------------------
# Count pale cells
# -------------------------------------------------
pale_count, pale_centers = count_cells(
    pale_mask,
    min_area=100
)


# -------------------------------------------------
# Print results
# -------------------------------------------------
print("Number of dark cells:", dark_count)
print("Number of pale cells:", pale_count)


# -------------------------------------------------
# Display dark cells
# -------------------------------------------------
dark_result = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

for i, (x, y) in enumerate(dark_centers, 1):

    cv2.circle(
        dark_result,
        (x, y),
        10,
        (0, 0, 255),
        2
    )

    cv2.putText(
        dark_result,
        str(i),
        (x + 10, y),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.5,
        (0, 0, 255),
        2
    )


# -------------------------------------------------
# Display pale cells
# -------------------------------------------------
pale_result = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

for i, (x, y) in enumerate(pale_centers, 1):

    cv2.circle(
        pale_result,
        (x, y),
        10,
        (0, 255, 0),
        2
    )

    cv2.putText(
        pale_result,
        str(i),
        (x + 10, y),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.5,
        (0, 255, 0),
        2
    )


# -------------------------------------------------
# Display results
# -------------------------------------------------
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(dark_mask, cmap="gray")
plt.title(f"Dark Cells Mask = {dark_count}")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(dark_result, cv2.COLOR_BGR2RGB))
plt.title(f"Dark Cells: {dark_count}")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(pale_result, cv2.COLOR_BGR2RGB))
plt.title(f"Pale Cells: {pale_count}")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# 1. Read image
# =========================================

image = cv2.imread(
    r"C:\Users\PC-LAB1\Desktop\New folder\bloodcells.jpeg"
)

if image is None:
    print("Image not found!")
    exit()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


# =========================================
# 2. Segmentation using Otsu
# =========================================

_, binary = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)


# =========================================
# 3. Remove noise
# =========================================

kernel = np.ones((3, 3), np.uint8)

opening = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    kernel,
    iterations=2
)


# =========================================
# 4. Sure Background
# =========================================

sure_bg = cv2.dilate(
    opening,
    kernel,
    iterations=3
)


# =========================================
# 5. Distance Transform
# =========================================

dist_transform = cv2.distanceTransform(
    opening,
    cv2.DIST_L2,
    5
)


# =========================================
# 6. Sure Foreground
# =========================================
# 0.4 يمكن تغييرها إلى 0.3 أو 0.5

_, sure_fg = cv2.threshold(
    dist_transform,
    0.4 * dist_transform.max(),
    255,
    0
)

sure_fg = np.uint8(sure_fg)


# =========================================
# 7. Unknown Region
# =========================================

unknown = cv2.subtract(
    sure_bg,
    sure_fg
)


# =========================================
# 8. Create markers
# =========================================

num_markers, markers = cv2.connectedComponents(
    sure_fg
)

print("Initial markers:", num_markers - 1)

# Background = 1
markers = markers + 1

# Unknown = 0
markers[unknown == 255] = 0


# =========================================
# 9. Watershed
# =========================================

markers = cv2.watershed(
    image,
    markers
)


# =========================================
# 10. Count cells
# =========================================

cell_count = 0

result = image.copy()

for label in range(2, num_markers + 1):

    mask = np.zeros(gray.shape, dtype=np.uint8)

    mask[markers == label] = 255

    area = cv2.countNonZero(mask)

    # Ignore very small regions
    if area > 100:

        cell_count += 1

        # Find center
        M = cv2.moments(mask)

        if M["m00"] != 0:

            cx = int(M["m10"] / M["m00"])
            cy = int(M["m01"] / M["m00"])

            # Draw center
            cv2.circle(
                result,
                (cx, cy),
                5,
                (0, 0, 255),
                -1
            )

            # Number cell
            cv2.putText(
                result,
                str(cell_count),
                (cx + 5, cy),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                2
            )


print("Total cells:", cell_count)


# =========================================
# 11. Display
# =========================================

plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary, cmap="gray")
plt.title("Otsu Segmentation")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(dist_transform, cmap="jet")
plt.title("Distance Transform")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(markers, cmap="nipy_spectral")
plt.title("Watershed")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title(f"Detected Cells = {cell_count}")
plt.axis("off")

plt.tight_layout()
plt.show()
```
