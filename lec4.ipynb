```python
import cv2

# Load the video
video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
else:
    # Video information
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps if fps > 0 else 0

    print("Video opened successfully!")
    print("-----------------------------")
    print(f"Width: {width} pixels")
    print(f"Height: {height} pixels")
    print(f"FPS: {fps:.2f}")
    print(f"Frame count: {frame_count}")
    print(f"Duration: {duration:.2f} seconds")
    print("-----------------------------")

    frame_number = 0

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        # ------------------------------------------------
        # 1. Grayscale conversion
        # ------------------------------------------------
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # ------------------------------------------------
        # 2. Gaussian filtering
        # ------------------------------------------------
        gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

        # Median filtering
        median = cv2.medianBlur(gray, 5)

        # ------------------------------------------------
        # 3. Canny edge detection
        # ------------------------------------------------
        canny = cv2.Canny(gaussian, 50, 150)

        # ------------------------------------------------
        # 4. Sobel edge detection
        # ------------------------------------------------
        sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
        sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

        sobel_x = cv2.convertScaleAbs(sobel_x)
        sobel_y = cv2.convertScaleAbs(sobel_y)

        sobel = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

        # ------------------------------------------------
        # 5. Scharr edge detection
        # ------------------------------------------------
        scharr_x = cv2.Scharr(gray, cv2.CV_64F, 1, 0)
        scharr_y = cv2.Scharr(gray, cv2.CV_64F, 0, 1)

        scharr_x = cv2.convertScaleAbs(scharr_x)
        scharr_y = cv2.convertScaleAbs(scharr_y)

        scharr = cv2.addWeighted(scharr_x, 0.5, scharr_y, 0.5, 0)

        # ------------------------------------------------
        # 6. Global thresholding
        # ------------------------------------------------
        _, global_thresh = cv2.threshold(
            gray, 127, 255, cv2.THRESH_BINARY
        )

        # ------------------------------------------------
        # 7. Otsu thresholding
        # ------------------------------------------------
        _, otsu = cv2.threshold(
            gray, 0, 255,
            cv2.THRESH_BINARY + cv2.THRESH_OTSU
        )

        # ------------------------------------------------
        # 8. Adaptive thresholding
        # ------------------------------------------------
        adaptive = cv2.adaptiveThreshold(
            gray,
            255,
            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY,
            11,
            2
        )

        # ------------------------------------------------
        # 9. Morphological operations
        # ------------------------------------------------
        kernel = cv2.getStructuringElement(
            cv2.MORPH_RECT, (5, 5)
        )

        # Erosion
        erosion = cv2.erode(gray, kernel, iterations=1)

        # Dilation
        dilation = cv2.dilate(gray, kernel, iterations=1)

        # Opening
        opening = cv2.morphologyEx(
            gray, cv2.MORPH_OPEN, kernel
        )

        # Closing
        closing = cv2.morphologyEx(
            gray, cv2.MORPH_CLOSE, kernel
        )

        # ------------------------------------------------
        # Display
        # ------------------------------------------------

        cv2.imshow("Original", frame)
        cv2.imshow("Grayscale", gray)
        cv2.imshow("Gaussian Blur", gaussian)
        cv2.imshow("Median Blur", median)
        cv2.imshow("Canny", canny)
        cv2.imshow("Sobel", sobel)
        cv2.imshow("Scharr", scharr)
        cv2.imshow("Global Threshold", global_thresh)
        cv2.imshow("Otsu", otsu)
        cv2.imshow("Adaptive Threshold", adaptive)
        cv2.imshow("Erosion", erosion)
        cv2.imshow("Dilation", dilation)
        cv2.imshow("Opening", opening)
        cv2.imshow("Closing", closing)

        # Print current frame
        print(f"Current frame: {frame_number}")
        frame_number += 1

        # ESC or q to stop
        key = cv2.waitKey(30) & 0xFF

        if key == 27 or key == ord('q'):
            break

# Release video
cap.release()
cv2.destroyAllWindows()


```

```python
import cv2

# قراءة الصورة
image = cv2.imread(r"FRAMES\frame_01412.jpg")

# 1. تحويل إلى Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. تقدير الخلفية وطرحها
background = cv2.GaussianBlur(gray, (51, 51), 0)
foreground = cv2.absdiff(gray, background)

# 3. Threshold
_, binary = cv2.threshold(
    foreground,
    30,
    255,
    cv2.THRESH_BINARY
)

# تنظيف الصورة من الضوضاء
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
binary = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
binary = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# 4. إيجاد الـ Contours
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# 5. حذف الأجسام الصغيرة
min_area = 500
count = 0

for contour in contours:

    if cv2.contourArea(contour) < min_area:
        continue

    # 6. رسم Bounding Rectangle
    x, y, w, h = cv2.boundingRect(contour)

    cv2.rectangle(
        image,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )

    count += 1

# 7. عرض عدد الأجسام
cv2.putText(
    image,
    f"Detected objects: {count}",
    (20, 40),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 0, 255),
    2
)

# عرض النتائج
cv2.imshow("Foreground", foreground)
cv2.imshow("Binary", binary)
cv2.imshow("Detected Objects", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
import math

# ------------------------------------------------
# Load the video
# ------------------------------------------------
video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()

# ------------------------------------------------
# Video information
# ------------------------------------------------
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0

print("Video opened successfully!")
print("-----------------------------")
print(f"Width: {width} pixels")
print(f"Height: {height} pixels")
print(f"FPS: {fps:.2f}")
print(f"Frame count: {frame_count}")
print(f"Duration: {duration:.2f} seconds")
print("-----------------------------")


# ------------------------------------------------
# Background Subtraction
# ------------------------------------------------
background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)


# ------------------------------------------------
# Object tracking variables
# ------------------------------------------------
next_object_id = 0

# Dictionary:
# ID -> (center_x, center_y)
objects = {}

# Maximum distance allowed between old and new center
max_distance = 50

# Minimum object area
min_area = 500


# ------------------------------------------------
# Function to calculate distance
# ------------------------------------------------
def calculate_distance(point1, point2):
    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# ------------------------------------------------
# Read video continuously
# ------------------------------------------------
frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # ------------------------------------------------
    # 1. Background Subtraction
    # ------------------------------------------------
    fg_mask = background_subtractor.apply(frame)

    # ------------------------------------------------
    # 2. Remove shadows and noise
    # ------------------------------------------------
    _, fg_mask = cv2.threshold(
        fg_mask,
        200,
        255,
        cv2.THRESH_BINARY
    )

    # ------------------------------------------------
    # 3. Morphological operations
    # ------------------------------------------------
    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    # Remove small noise
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )

    # Connect nearby regions
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_CLOSE,
        kernel
    )

    # Expand objects slightly
    fg_mask = cv2.dilate(
        fg_mask,
        kernel,
        iterations=2
    )


    # ------------------------------------------------
    # 4. Find contours
    # ------------------------------------------------
    contours, _ = cv2.findContours(
        fg_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # ------------------------------------------------
    # Store detected objects in current frame
    # ------------------------------------------------
    current_objects = []

    for contour in contours:

        # Calculate area
        area = cv2.contourArea(contour)

        # Ignore small objects/noise
        if area < min_area:
            continue

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Center of object
        center_x = x + w // 2
        center_y = y + h // 2

        current_objects.append(
            (center_x, center_y, x, y, w, h)
        )


    # ------------------------------------------------
    # 5. Object Tracking
    # ------------------------------------------------

    new_objects = {}

    # Keep track of IDs that have already been used
    used_ids = set()

    for center_x, center_y, x, y, w, h in current_objects:

        current_center = (center_x, center_y)

        best_id = None
        best_distance = max_distance

        # Compare with previous objects
        for object_id, previous_center in objects.items():

            if object_id in used_ids:
                continue

            distance = calculate_distance(
                current_center,
                previous_center
            )

            if distance < best_distance:
                best_distance = distance
                best_id = object_id

        # If no matching object was found,
        # create a new ID
        if best_id is None:

            best_id = next_object_id
            next_object_id += 1

        # Save object
        new_objects[best_id] = current_center
        used_ids.add(best_id)

        # ------------------------------------------------
        # 6. Draw Bounding Box
        # ------------------------------------------------
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Draw center point
        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )

        # Display object ID
        cv2.putText(
            frame,
            f"ID: {best_id}",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )


    # Update objects
    objects = new_objects


    # ------------------------------------------------
    # 7. Display frame number
    # ------------------------------------------------
    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # Number of detected objects
    cv2.putText(
        frame,
        f"Objects: {len(current_objects)}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # ------------------------------------------------
    # 8. Display results
    # ------------------------------------------------

    cv2.imshow("Object Detection and Tracking", frame)

    cv2.imshow("Background Subtraction", fg_mask)


    print(
        f"Frame: {frame_number} | "
        f"Objects detected: {len(current_objects)}"
    )

    frame_number += 1


    # ------------------------------------------------
    # 9. ESC or q to stop
    # ------------------------------------------------
    key = cv2.waitKey(30) & 0xFF

    if key == 27 or key == ord('q'):
        break


# ------------------------------------------------
# Release everything
# ------------------------------------------------
cap.release()
cv2.destroyAllWindows()

print("Finished.")

```

```python

import cv2
import math
import os

# =========================================================
# 1. Load Video
# =========================================================

video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()

# =========================================================
# 2. Video Information
# =========================================================

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0

print("Video opened successfully!")
print("--------------------------------")
print(f"Width       : {width} pixels")
print(f"Height      : {height} pixels")
print(f"FPS         : {fps:.2f}")
print(f"Frame Count : {frame_count}")
print(f"Duration    : {duration:.2f} seconds")
print("--------------------------------")


# =========================================================
# 3. Create Output Folder
# =========================================================

os.makedirs("OUTPUT", exist_ok=True)


# =========================================================
# 4. Create Output Video
# =========================================================

output_path = "OUTPUT/detected_cars2.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)


# =========================================================
# 5. Background Subtraction
# =========================================================

background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)


# =========================================================
# 6. Object Tracking Variables
# =========================================================

next_object_id = 0

objects = {}

# IDs of cars that have been counted
counted_ids = set()

# Maximum distance between old and new center
max_distance = 60

# Minimum area for a detected object
min_area = 800


# =========================================================
# 7. Distance Function
# =========================================================

def calculate_distance(point1, point2):

    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# =========================================================
# 8. Frame Processing
# =========================================================

frame_number = 0


while True:

    ret, frame = cap.read()

    if not ret:
        break


    # =====================================================
    # A. Grayscale
    # =====================================================

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )


    # =====================================================
    # B. Gaussian Filtering
    # =====================================================

    gaussian = cv2.GaussianBlur(
        gray,
        (5, 5),
        0
    )


    # =====================================================
    # C. Median Filtering
    # =====================================================

    median = cv2.medianBlur(
        gray,
        5
    )


    # =====================================================
    # D. Canny Edge Detection
    # =====================================================

    canny = cv2.Canny(
        gaussian,
        50,
        150
    )


    # =====================================================
    # E. Sobel
    # =====================================================

    sobel_x = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    sobel_y = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    sobel_x = cv2.convertScaleAbs(sobel_x)
    sobel_y = cv2.convertScaleAbs(sobel_y)

    sobel = cv2.addWeighted(
        sobel_x,
        0.5,
        sobel_y,
        0.5,
        0
    )


    # =====================================================
    # F. Scharr
    # =====================================================

    scharr_x = cv2.Scharr(
        gray,
        cv2.CV_64F,
        1,
        0
    )

    scharr_y = cv2.Scharr(
        gray,
        cv2.CV_64F,
        0,
        1
    )

    scharr_x = cv2.convertScaleAbs(scharr_x)
    scharr_y = cv2.convertScaleAbs(scharr_y)

    scharr = cv2.addWeighted(
        scharr_x,
        0.5,
        scharr_y,
        0.5,
        0
    )


    # =====================================================
    # G. Global Threshold
    # =====================================================

    _, global_thresh = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )


    # =====================================================
    # H. Otsu Threshold
    # =====================================================

    _, otsu = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )


    # =====================================================
    # I. Adaptive Threshold
    # =====================================================

    adaptive = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11,
        2
    )


    # =====================================================
    # J. Background Subtraction
    # =====================================================

    fg_mask = background_subtractor.apply(frame)


    # =====================================================
    # K. Remove Shadows
    # =====================================================

    _, fg_mask = cv2.threshold(
        fg_mask,
        200,
        255,
        cv2.THRESH_BINARY
    )


    # =====================================================
    # L. Morphological Operations
    # =====================================================

    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )


    # Opening
    opening = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )


    # Closing
    closing = cv2.morphologyEx(
        opening,
        cv2.MORPH_CLOSE,
        kernel
    )


    # Dilation
    dilation = cv2.dilate(
        closing,
        kernel,
        iterations=2
    )


    # Erosion
    erosion = cv2.erode(
        dilation,
        kernel,
        iterations=1
    )


    # =====================================================
    # M. Find Contours
    # =====================================================

    contours, _ = cv2.findContours(
        erosion,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    current_objects = []


    for contour in contours:

        area = cv2.contourArea(contour)

        # Ignore small objects
        if area < min_area:
            continue


        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)


        # Object center
        center_x = x + w // 2
        center_y = y + h // 2


        current_objects.append(
            (
                center_x,
                center_y,
                x,
                y,
                w,
                h
            )
        )


    # =====================================================
    # N. Track Objects
    # =====================================================

    new_objects = {}

    used_ids = set()


    for center_x, center_y, x, y, w, h in current_objects:

        current_center = (
            center_x,
            center_y
        )


        best_id = None
        best_distance = max_distance


        # Compare with previous frame
        for object_id, previous_center in objects.items():

            if object_id in used_ids:
                continue


            distance = calculate_distance(
                current_center,
                previous_center
            )


            if distance < best_distance:

                best_distance = distance
                best_id = object_id


        # New car
        if best_id is None:

            best_id = next_object_id

            next_object_id += 1

            # Add to total counted cars
            counted_ids.add(best_id)


        # Save object
        new_objects[best_id] = current_center

        used_ids.add(best_id)


        # =================================================
        # Draw Bounding Box
        # =================================================

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )


        # =================================================
        # Draw Center
        # =================================================

        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )


        # =================================================
        # Display ID
        # =================================================

        cv2.putText(
            frame,
            f"Car ID: {best_id}",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )


    # Update tracked objects
    objects = new_objects


    # =====================================================
    # O. Display Frame Number
    # =====================================================

    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 0),
        2
    )


    # =====================================================
    # P. Display Total Cars
    # =====================================================

    total_cars = len(counted_ids)

    cv2.putText(
        frame,
        f"Total Cars: {total_cars}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 255),
        2
    )


    # =====================================================
    # Q. Display Processing Results
    # =====================================================

    cv2.imshow(
        "Original + Detection",
        frame
    )

    cv2.imshow(
        "1 - Grayscale",
        gray
    )

    cv2.imshow(
        "2 - Gaussian",
        gaussian
    )

    cv2.imshow(
        "3 - Median",
        median
    )

    cv2.imshow(
        "4 - Canny",
        canny
    )

    cv2.imshow(
        "5 - Sobel",
        sobel
    )

    cv2.imshow(
        "6 - Scharr",
        scharr
    )

    cv2.imshow(
        "7 - Global Threshold",
        global_thresh
    )

    cv2.imshow(
        "8 - Otsu",
        otsu
    )

    cv2.imshow(
        "9 - Adaptive Threshold",
        adaptive
    )

    cv2.imshow(
        "10 - Background Subtraction",
        fg_mask
    )

    cv2.imshow(
        "11 - Opening",
        opening
    )

    cv2.imshow(
        "12 - Closing",
        closing
    )

    cv2.imshow(
        "13 - Dilation",
        dilation
    )

    cv2.imshow(
        "14 - Erosion",
        erosion
    )


    # =====================================================
    # R. Save Processed Video
    # =====================================================

    out.write(frame)


    # Print information
    print(
        f"Frame: {frame_number} | "
        f"Cars detected: {len(current_objects)} | "
        f"Total cars: {total_cars}"
    )


    frame_number += 1


    # =====================================================
    # S. ESC or Q to Stop
    # =====================================================

    key = cv2.waitKey(30) & 0xFF

    if key == 27 or key == ord('q'):
        break


# =========================================================
# 9. Release Everything
# =========================================================

cap.release()
out.release()

cv2.destroyAllWindows()


# =========================================================
# 10. Final Result
# =========================================================

print("--------------------------------")
print("Processing finished!")
print(f"Total cars detected: {len(counted_ids)}")
print(f"Output video saved to: {output_path}")
print("--------------------------------")

```

```python
import cv2
import math
import os

# =========================================================
# 1. Load Video
# =========================================================

video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()

# =========================================================
# 2. Video Information
# =========================================================

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0

print("Video opened successfully!")
print("--------------------------------")
print(f"Width       : {width} pixels")
print(f"Height      : {height} pixels")
print(f"FPS         : {fps:.2f}")
print(f"Frame Count : {frame_count}")
print(f"Duration    : {duration:.2f} seconds")
print("--------------------------------")


# =========================================================
# 3. Measurement Settings
# =========================================================

# Real distance between the two lines
REAL_DISTANCE = 8.0       # meters

# Measurement lines
# These values can be adjusted according to your video
LINE1_Y = 250
LINE2_Y = 500


# =========================================================
# 4. Output Video
# =========================================================

os.makedirs("OUTPUT", exist_ok=True)

output_path = "OUTPUT/detected_cars_speed.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")

out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)


# =========================================================
# 5. Background Subtraction
# =========================================================

background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)


# =========================================================
# 6. Tracking Variables
# =========================================================

next_object_id = 0

objects = {}

# Information for each car
car_data = {}

# Cars that crossed the first line
cars_started = set()

# Cars whose speed has been calculated
cars_finished = set()

# Total cars that passed the measurement area
counted_cars = set()

# Maximum distance for matching objects
MAX_DISTANCE = 70

# Minimum contour area
MIN_AREA = 800


# =========================================================
# 7. Distance Function
# =========================================================

def calculate_distance(point1, point2):

    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# =========================================================
# 8. Main Loop
# =========================================================

frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break


    # =====================================================
    # IMAGE PROCESSING
    # =====================================================

    # -----------------------------------------------------
    # 1. Grayscale
    # -----------------------------------------------------

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )


    # -----------------------------------------------------
    # 2. Gaussian Blur
    # -----------------------------------------------------

    gaussian = cv2.GaussianBlur(
        gray,
        (5, 5),
        0
    )


    # -----------------------------------------------------
    # 3. Median Filter
    # -----------------------------------------------------

    median = cv2.medianBlur(
        gray,
        5
    )


    # -----------------------------------------------------
    # 4. Canny
    # -----------------------------------------------------

    canny = cv2.Canny(
        gaussian,
        50,
        150
    )


    # -----------------------------------------------------
    # 5. Sobel
    # -----------------------------------------------------

    sobel_x = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    sobel_y = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    sobel_x = cv2.convertScaleAbs(sobel_x)
    sobel_y = cv2.convertScaleAbs(sobel_y)

    sobel = cv2.addWeighted(
        sobel_x,
        0.5,
        sobel_y,
        0.5,
        0
    )


    # -----------------------------------------------------
    # 6. Scharr
    # -----------------------------------------------------

    scharr_x = cv2.Scharr(
        gray,
        cv2.CV_64F,
        1,
        0
    )

    scharr_y = cv2.Scharr(
        gray,
        cv2.CV_64F,
        0,
        1
    )

    scharr_x = cv2.convertScaleAbs(scharr_x)
    scharr_y = cv2.convertScaleAbs(scharr_y)

    scharr = cv2.addWeighted(
        scharr_x,
        0.5,
        scharr_y,
        0.5,
        0
    )


    # -----------------------------------------------------
    # 7. Global Threshold
    # -----------------------------------------------------

    _, global_thresh = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )


    # -----------------------------------------------------
    # 8. Otsu
    # -----------------------------------------------------

    _, otsu = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )


    # -----------------------------------------------------
    # 9. Adaptive Threshold
    # -----------------------------------------------------

    adaptive = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        11,
        2
    )


    # =====================================================
    # BACKGROUND SUBTRACTION
    # =====================================================

    fg_mask = background_subtractor.apply(frame)


    # Remove shadows
    _, fg_mask = cv2.threshold(
        fg_mask,
        200,
        255,
        cv2.THRESH_BINARY
    )


    # =====================================================
    # MORPHOLOGICAL OPERATIONS
    # =====================================================

    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )


    # Opening
    opening = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )


    # Closing
    closing = cv2.morphologyEx(
        opening,
        cv2.MORPH_CLOSE,
        kernel
    )


    # Dilation
    dilation = cv2.dilate(
        closing,
        kernel,
        iterations=2
    )


    # Erosion
    erosion = cv2.erode(
        dilation,
        kernel,
        iterations=1
    )


    # =====================================================
    # FIND OBJECTS
    # =====================================================

    contours, _ = cv2.findContours(
        erosion,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    current_objects = []


    for contour in contours:

        area = cv2.contourArea(contour)

        if area < MIN_AREA:
            continue


        x, y, w, h = cv2.boundingRect(contour)


        center_x = x + w // 2
        center_y = y + h // 2


        current_objects.append(
            (
                center_x,
                center_y,
                x,
                y,
                w,
                h
            )
        )


    # =====================================================
    # TRACKING
    # =====================================================

    new_objects = {}

    used_ids = set()


    for center_x, center_y, x, y, w, h in current_objects:

        current_center = (
            center_x,
            center_y
        )


        best_id = None
        best_distance = MAX_DISTANCE


        # Find nearest previous object
        for object_id, previous_center in objects.items():

            if object_id in used_ids:
                continue


            distance = calculate_distance(
                current_center,
                previous_center
            )


            if distance < best_distance:

                best_distance = distance
                best_id = object_id


        # -------------------------------------------------
        # New object
        # -------------------------------------------------

        if best_id is None:

            best_id = next_object_id

            next_object_id += 1


            car_data[best_id] = {
                "start_frame": None,
                "speed": None,
                "last_y": center_y
            }


        # -------------------------------------------------
        # Save object
        # -------------------------------------------------

        new_objects[best_id] = current_center

        used_ids.add(best_id)


        # =================================================
        # SPEED CALCULATION
        # =================================================

        previous_y = car_data[best_id]["last_y"]


        # -------------------------------------------------
        # Car crosses LINE 1
        # -------------------------------------------------

        if (
            previous_y < LINE1_Y
            and center_y >= LINE1_Y
            and best_id not in cars_started
        ):

            car_data[best_id]["start_frame"] = frame_number

            cars_started.add(best_id)


            print(
                f"Car {best_id} crossed LINE 1"
            )


        # -------------------------------------------------
        # Car crosses LINE 2
        # -------------------------------------------------

        if (
            previous_y < LINE2_Y
            and center_y >= LINE2_Y
            and best_id in cars_started
            and best_id not in cars_finished
        ):

            start_frame = car_data[best_id]["start_frame"]

            # Number of frames
            frames_taken = frame_number - start_frame


            # Time in seconds
            time_seconds = frames_taken / fps


            if time_seconds > 0:

                # Speed m/s
                speed_mps = REAL_DISTANCE / time_seconds


                # Speed km/h
                speed_kmh = speed_mps * 3.6


                car_data[best_id]["speed"] = speed_kmh


                cars_finished.add(best_id)

                counted_cars.add(best_id)


                print(
                    f"Car {best_id} speed = "
                    f"{speed_kmh:.2f} km/h"
                )


        # Update previous Y
        car_data[best_id]["last_y"] = center_y


        # =================================================
        # DRAW CAR
        # =================================================

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )


        # Center
        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )


        # =================================================
        # DISPLAY ID
        # =================================================

        cv2.putText(
            frame,
            f"Car ID: {best_id}",
            (x, y - 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )


        # =================================================
        # DISPLAY SPEED
        # =================================================

        speed = car_data[best_id]["speed"]

        if speed is not None:

            cv2.putText(
                frame,
                f"Speed: {speed:.1f} km/h",
                (x, y - 8),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 255),
                2
            )


    # Update objects
    objects = new_objects


    # =====================================================
    # DRAW MEASUREMENT LINES
    # =====================================================

    cv2.line(
        frame,
        (0, LINE1_Y),
        (width, LINE1_Y),
        (255, 0, 0),
        3
    )

    cv2.line(
        frame,
        (0, LINE2_Y),
        (width, LINE2_Y),
        (0, 0, 255),
        3
    )


    # Line labels
    cv2.putText(
        frame,
        "LINE 1",
        (20, LINE1_Y - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 0, 0),
        2
    )


    cv2.putText(
        frame,
        "LINE 2 - 8 meters",
        (20, LINE2_Y - 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 0, 255),
        2
    )


    # =====================================================
    # DISPLAY FRAME NUMBER
    # =====================================================

    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 0),
        2
    )


    # =====================================================
    # TOTAL CARS
    # =====================================================

    total_cars = len(counted_cars)

    cv2.putText(
        frame,
        f"Total Cars: {total_cars}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 255),
        2
    )


    # =====================================================
    # DISPLAY MAIN VIDEO
    # =====================================================

    cv2.imshow(
        "Cars Detection + Tracking + Speed",
        frame
    )


    # =====================================================
    # DISPLAY PROCESSING
    # =====================================================

    cv2.imshow(
        "Grayscale",
        gray
    )

    cv2.imshow(
        "Gaussian Blur",
        gaussian
    )

    cv2.imshow(
        "Median Blur",
        median
    )

    cv2.imshow(
        "Canny",
        canny
    )

    cv2.imshow(
        "Sobel",
        sobel
    )

    cv2.imshow(
        "Scharr",
        scharr
    )

    cv2.imshow(
        "Global Threshold",
        global_thresh
    )

    cv2.imshow(
        "Otsu",
        otsu
    )

    cv2.imshow(
        "Adaptive Threshold",
        adaptive
    )

    cv2.imshow(
        "Background Subtraction",
        fg_mask
    )

    cv2.imshow(
        "Opening",
        opening
    )

    cv2.imshow(
        "Closing",
        closing
    )

    cv2.imshow(
        "Dilation",
        dilation
    )

    cv2.imshow(
        "Erosion",
        erosion
    )


    # =====================================================
    # SAVE VIDEO
    # =====================================================

    out.write(frame)


    # Console
    print(
        f"Frame: {frame_number} | "
        f"Current objects: {len(current_objects)} | "
        f"Cars completed: {total_cars}"
    )


    frame_number += 1


    # =====================================================
    # ESC or Q
    # =====================================================

    key = cv2.waitKey(30) & 0xFF

    if key == 27 or key == ord("q"):
        break


# =========================================================
# RELEASE
# =========================================================

cap.release()
out.release()

cv2.destroyAllWindows()


# =========================================================
# FINAL RESULT
# =========================================================

print()
print("======================================")
print("Processing Finished!")
print(f"Total Cars: {len(counted_cars)}")
print(f"Output Video: {output_path}")
print("======================================")
```
